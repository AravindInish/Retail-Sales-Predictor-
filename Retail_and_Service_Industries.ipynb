```python
import kagglehub
path = kagglehub.dataset_download("arunjangir245/retail-and-service-industries-dataset")
```

```python
import os

# List the files in the downloaded dataset directory
file_list = os.listdir(path)
print(f"Files in the dataset directory: {file_list}")
```

```python
import pandas as pd

# Construct the full path to the CSV file
csv_file_path = os.path.join(path, file_list[0])

# Load the CSV file into a pandas DataFrame
df = pd.read_csv(csv_file_path)

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Get information about the DataFrame (data types, non-null values)
df.info()

# Get descriptive statistics for numerical columns
df.describe()
```

```python
# Convert 'month' column to datetime objects
df['month'] = pd.to_datetime(df['month'], format='%d-%m-%Y')

# Check for duplicate rows
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

# Display the DataFrame info again to confirm datetime conversion
df.info()
```

```python
import matplotlib.pyplot as plt

# Set 'month' as the DataFrame index
df = df.set_index('month')

# Sort the DataFrame by index to ensure chronological order
df = df.sort_index()

# Plot the 'supermarket_and_grocery_stores' column
plt.figure(figsize=(15, 6))
plt.plot(df.index, df['supermarket_and_grocery_stores'], label='Supermarket and Grocery Stores Sales')
plt.title('Supermarket and Grocery Stores Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()

print("DataFrame after setting 'month' as index:")
display(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

# Select numerical columns for scaling (all columns except the index 'month')
numerical_cols = df.columns

# Initialize the MinMaxScaler
scaler = MinMaxScaler()

# Fit and transform the data
df_scaled = pd.DataFrame(scaler.fit_transform(df[numerical_cols]), columns=numerical_cols, index=df.index)

print("Scaled DataFrame head:")
display(df_scaled.head())
print("Scaled DataFrame describe:")
display(df_scaled.describe())
```

```python
import numpy as np

def create_sequences(data, sequence_length):
    X, y = [], []
    for i in range(len(data) - sequence_length):
        X.append(data[i:i + sequence_length])
        y.append(data[i + sequence_length])
    return np.array(X), np.array(y)

# Define the sequence length (look-back period)
sequence_length = 12  # For example, use the past 12 months to predict the next

# Select the target column for prediction (e.g., 'supermarket_and_grocery_stores')
target_column = 'supermarket_and_grocery_stores'

# Prepare the data for sequence creation
data_for_sequences = df_scaled[target_column].values.reshape(-1, 1)

# Create sequences
X, y = create_sequences(data_for_sequences, sequence_length)

print(f"Shape of X (input sequences): {X.shape}")
print(f"Shape of y (target values): {y.shape}")

print("First sample of X:")
print(X[0])
print("First sample of y:")
print(y[0])
```

```python
# Split data into training and testing sets
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# Build the LSTM model
model = Sequential([
    LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1], 1)),
    Dropout(0.2),
    LSTM(units=50, return_sequences=False),
    Dropout(0.2),
    Dense(units=1) # Output layer for predicting the next single value
])

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error')

# Print model summary
model.summary()
```

```python
# Train the model
history = model.fit(
    X_train,
    y_train,
    epochs=50, # Number of training iterations
    batch_size=32, # Number of samples per gradient update
    validation_data=(X_test, y_test), # Data for validation during training
    verbose=1 # Show progress bar
)

print("Model training complete.")
```

```python
import matplotlib.pyplot as plt

# Plot training and validation loss
plt.figure(figsize=(12, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss (Mean Squared Error)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import joblib

# Save the model
model.save('lstm_retail_sales_model.keras')
print("Model saved as 'lstm_retail_sales_model.keras'")

# Save the scaler
joblib.dump(scaler, 'minmax_scaler.pkl')
print("Scaler saved as 'minmax_scaler.pkl'")
```

```python
import joblib

# Save the numerical column names so the Streamlit app knows the order for inverse scaling
joblib.dump(numerical_cols, 'numerical_columns.pkl')
print("Numerical column names saved as 'numerical_columns.pkl'")
```

```python
%%writefile streamlit_app.py
import streamlit as st
import numpy as np
import pandas as pd
import joblib
from tensorflow.keras.models import load_model

# Load the trained model and scaler
model = load_model('lstm_retail_sales_model.keras')
scaler = joblib.load('minmax_scaler.pkl')

# Define the target column used during training (needed for inverse scaling)
target_column = 'supermarket_and_grocery_stores'

st.set_page_config(
    page_title="Retail Sales Predictor 📈",
    page_icon="📊",
    layout="centered",
    initial_sidebar_state="expanded",
)

st.title("Retail Sales Predictor 📈")
st.write("This app predicts the next month's 'Supermarket and Grocery Stores' sales using a trained LSTM model. Please provide the past 12 months of *scaled* sales data in the sidebar.")

# Sidebar for user input
st.sidebar.header("Input Past 12 Months Scaled Sales Data 👇")
st.sidebar.write("Enter scaled values (between 0 and 1) for the last 12 months.")

input_sequence = []
for i in range(12):
    value = st.sidebar.number_input(f"Month {i+1} Scaled Sales", min_value=0.0, max_value=1.0, value=0.5, step=0.01)
    input_sequence.append(value)

predict_button = st.sidebar.button("Predict Next Month's Sales 🔮")

st.subheader("Prediction Results ✨")

if predict_button:
    if len(input_sequence) == 12:
        # Reshape input for the model: (1, 12, 1)
        input_array = np.array(input_sequence).reshape(1, 12, 1)

        # Make prediction
        scaled_prediction = model.predict(input_array)[0, 0]

        # Create a dummy array for inverse transformation
        # The scaler was fitted on the entire df_scaled, so it expects a 2D array with all columns
        # We need to construct a row that looks like the original data before scaling
        # to inverse transform only the target column accurately.
        # This requires knowing the original column order and putting the scaled_prediction
        # in the correct position.

        # The simplest way to inverse transform a single value from a MinMaxScaler
        # trained on multiple columns is to create an array with the single scaled
        # value at the position of the target column, and other columns can be anything (e.g., 0).
        # Then inverse transform this full array and extract the target column's value.

        # Get the index of the target column
        target_col_index = df_scaled.columns.get_loc(target_column)

        # Create a zero array with the number of features the scaler was trained on
        dummy_row = np.zeros((1, scaler.n_features_in_))

        # Place the scaled prediction in the target column's position
        dummy_row[0, target_col_index] = scaled_prediction

        # Inverse transform the dummy row
        original_scale_prediction = scaler.inverse_transform(dummy_row)[0, target_col_index]

        st.success(f"The predicted next month's sales for '{target_column}' are: **${original_scale_prediction:,.2f}** 💰")
    else:
        st.error("Please enter 12 scaled sales values.")
else:
    st.info("Enter the scaled sales data for the last 12 months in the sidebar and click 'Predict' to see the forecast! 👆")
```

```python
%%writefile streamlit_app.py
import streamlit as st
import numpy as np
import pandas as pd # Import pandas to use Index.get_loc
import joblib
from tensorflow.keras.models import load_model

# Load the trained model and scaler
model = load_model('lstm_retail_sales_model.keras')
scaler = joblib.load('minmax_scaler.pkl')

# Load the numerical column names for inverse transformation context
numerical_cols = joblib.load('numerical_columns.pkl')

# Define the target column for prediction
target_column = 'supermarket_and_grocery_stores'

st.set_page_config(
    page_title="Retail Sales Predictor 📈",
    page_icon="📊",
    layout="centered",
    initial_sidebar_state="expanded",
)

st.title("Retail Sales Predictor 📈")
st.write("This app predicts the next month's 'Supermarket and Grocery Stores' sales using a trained LSTM model. Please provide the past 12 months of *scaled* sales data in the sidebar.")

# Sidebar for user input
st.sidebar.header("Input Past 12 Months Scaled Sales Data 👇")
st.sidebar.write("Enter scaled values (between 0 and 1) for the last 12 months.")

input_sequence = []
for i in range(12):
    # Ensure the default value is within the [0,1] range for scaled data
    value = st.sidebar.number_input(f"Month {i+1} Scaled Sales", min_value=0.0, max_value=1.0, value=0.5, step=0.01)
    input_sequence.append(value)

predict_button = st.sidebar.button("Predict Next Month's Sales 🔮")

st.subheader("Prediction Results ✨")

if predict_button:
    if len(input_sequence) == 12:
        # Reshape input for the model: (1, 12, 1)
        input_array = np.array(input_sequence).reshape(1, 12, 1)

        # Make prediction
        scaled_prediction = model.predict(input_array)[0, 0]

        # Get the index of the target column from the loaded numerical_cols
        target_col_index = pd.Index(numerical_cols).get_loc(target_column)

        # Create a zero array with the number of features the scaler was trained on
        dummy_row = np.zeros((1, scaler.n_features_in_))

        # Place the scaled prediction in the target column's position
        dummy_row[0, target_col_index] = scaled_prediction

        # Inverse transform the dummy row
        original_scale_prediction = scaler.inverse_transform(dummy_row)[0, target_col_index]

        st.success(f"The predicted next month's sales for '{target_column}' are: **${original_scale_prediction:,.2f}** 💰")
    else:
        st.error("Please enter 12 scaled sales values.")
else:
    st.info("Enter the scaled sales data for the last 12 months in the sidebar and click 'Predict' to see the forecast! 👆")
```

```python
# Make predictions on the test data
test_predictions_scaled = model.predict(X_test)

# Inverse transform the predictions and actual values to their original scale
# To inverse transform, we need to provide a dummy array with the correct number of features (columns)
# as the scaler was fitted on the entire df_scaled.

# Get the index of the target column
target_col_index = df_scaled.columns.get_loc(target_column)

# Create dummy arrays for inverse transformation
# For predictions
dummy_predictions_array = np.zeros((len(test_predictions_scaled), scaler.n_features_in_))
dummy_predictions_array[:, target_col_index] = test_predictions_scaled.flatten()
test_predictions = scaler.inverse_transform(dummy_predictions_array)[:, target_col_index]

# For actual values (y_test)
dummy_y_test_array = np.zeros((len(y_test), scaler.n_features_in_))
dummy_y_test_array[:, target_col_index] = y_test.flatten()
actual_values = scaler.inverse_transform(dummy_y_test_array)[:, target_col_index]

print("Predictions and actual values inverse transformed.")
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Calculate evaluation metrics
rmse = np.sqrt(mean_squared_error(actual_values, test_predictions))
mae = mean_absolute_error(actual_values, test_predictions)

print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"Mean Absolute Error (MAE): {mae:.2f}")

# Create a DataFrame for easier plotting and comparison
test_results = pd.DataFrame({'Actual': actual_values.flatten(), 'Predicted': test_predictions.flatten()})

print("\nFirst 5 test results:")
display(test_results.head())
```

```python
import matplotlib.pyplot as plt

# Visualize the predictions vs actuals for the test set
plt.figure(figsize=(15, 7))
plt.plot(test_results.index, test_results['Actual'], label='Actual Sales', color='blue')
plt.plot(test_results.index, test_results['Predicted'], label='Predicted Sales', color='red', linestyle='--')
plt.title(f'LSTM Model Predictions vs. Actuals (Test Set) - {target_column}')
plt.xlabel('Time Step (in Test Set)')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()

print("Plot of Actual vs. Predicted sales generated.")
```
